A GRU has 3 internal gates:

update gate $z_t$

reset gate $r_t$

candidate hidden state $h̃_t$

For a GRU with input dimension $D = 64$ and $U = 128$ units, using the default Keras setting reset_after=True, the parameter count is:

3 × (D × U + U × U + U) + 3 x U
= 3 × (64 × 128 + 128 × 128 + 128) + 3 x 128
= 3 × (8192 + 16384 + 128 + 128)
= 3 × 24832
= 74,496

## update gate
$z_t = \sigma( x_t W_z + h_{t-1} U_z + b_{input,z} + b_{rec,z} )$          

## reset gate
$r_t = \sigma ( x_t W_r + h_{t-1} U_r + b_{input,r} + b_{rec,r} )$          

## candidate
$\tilde{h}_t  = tanh( x_t W_h + b_{input,h} + r_t ⊙ (h_{t-1} U_h + b_{rec,h}) )$  

## new hidden state
$h_t = (1 - z_t) ⊙ h_{t-1} + z_t ⊙ \tilde{h}_t$         

Where:

$x_t$ is the input at time $t$, dimension $D$

$h_{t-1}$ is the previous hidden state, dimension $U$

$W_z, W_r, W_h$ are input weight matrices of shape $(D, U)$

$U_z, U_r, U_h$ are recurrent weight matrices of shape $(U, U)$

$b_{input,z}, b_{input,r}, b_{input,h}$ are bias vectors of shape $(U,)$

$b_{rec,z}, b_{rec,r}, b_{rec,h}$ are recurrent bias vectors of shape $(U,)$

$σ$ is the sigmoid function

$⊙$ is element-wise multiplication

```python
        x_t                     h_{t-1}
         │                         │
         ▼                         ▼
   ┌──────────┐              ┌──────────┐
   │ x_t W_z  │              │ h_{t-1} U_z │
   └────┬─────┘              └─────┬──────┘
        │ + b_input_z              │ + b_rec_z
        └───────────┬──────────────┘
                    ▼
                   σ  ──────────►  z_t

        x_t                     h_{t-1}
         │                         │
         ▼                         ▼
   ┌──────────┐              ┌──────────┐
   │ x_t W_r  │              │ h_{t-1} U_r │
   └────┬─────┘              └─────┬──────┘
        │ + b_input_r              │ + b_rec_r
        └───────────┬──────────────┘
                    ▼
                   σ  ──────────►  r_t

        x_t                     h_{t-1}
         │                         │
         ▼                         ▼
   ┌──────────┐              ┌──────────┐
   │ x_t W_h  │              │ h_{t-1} U_h │
   └────┬─────┘              └─────┬──────┘
        │ + b_input_h              │ + b_rec_h
        │                          ▼
        │                    ┌───────────┐
        │                    │  ⊙ r_t    │
        │                    └─────┬─────┘
        └───────────┬──────────────┘
                    ▼
                   tanh ────────►  h̃_t

   h_t = (1 - z_t) ⊙ h_{t-1} + z_t ⊙ h̃_t
```

```python
def get_train():
    seq = [[0.0, 0.1], [0.1,0.2], [0.2,0.3], [0.3,0.4], [0.4,0.5]]
    seq = array(seq)
    X,y= seq[:,0], seq[:,1]
    X = X.reshape((len(X),1,1))
    return X,y
```

```python
def get_val():
    seq = [[0.5, 0.6], [0.6,0.7], [0.7,0.8], [0.8,0.9], [0.9,1.0]]
    seq = array(seq)
    X,y= seq[:,0], seq[:,1]
    X = X.reshape((len(X),1,1))
    return X,y
    
```

```python
model = Sequential()

model.add(LSTM(10,input_shape=(1,1)))


model.add(Dense(1,activation='linear'))

model.summary()
```

```python
model.compile(loss='mse', optimizer='adam')
```

```python
X,y = get_train()

valX, valY = get_val()
```

```python
valX,valY
```
