```python
# Load the CSV file into a DataFrame
df = pd.read_csv('finalNew.csv')

# View the first 5 rows
print(df.head())
df
```

```python
df.index= pd.date_range(start='2023-05-01', periods=len(df), freq='M')
df = df[['Value']]
df
```

```python
df.info()
```

```python
# 3. Resample (e.g., daily 'D' or monthly 'ME') and apply an aggregation like .mean() or .sum()
df_resampled = df.resample('M').mean()
```

```python
df_resampled
```

```python
df_resampled['Value'].ffill()
```

```python
df_resampled['Value'] = df_resampled['Value'].ffill()
```

```python
df_resampled['Value'].isna().sum()
```

```python
df_resampled['Value']
```

```python
from pycaret.time_series import TSForecastingExperiment

# 1. Load your data (Ensure index is a DatetimeIndex or PeriodIndex with correct frequency)


# 2. Instantiate and setup the experiment
# 2. Instantiate and setup the experiment
exp = TSForecastingExperiment()
exp.setup(
    data=df_resampled,
    target='Value',
    fh=6,
    fold=3,
    seasonal_period=12,
    session_id=123
)
```

```python
# 4. Compare available models to select the top performer
# PyCaret evaluates models using metrics like MASE, RMSE, and MAPE
best_model = exp.compare_models()

# 5. Optional: Tune the selected best model to optimize hyperparameters
tuned_model = exp.tune_model(best_model)

# 6. Generate predictions for the next 12 months
future_forecast = exp.predict_model(tuned_model)
print("\n--- 6-Month Forecast ---")
print(future_forecast)

# 7. Finalize and save the model for deployment
final_model = exp.finalize_model(tuned_model)
exp.save_model(final_model, 'my_sales_forecast_model')
```
